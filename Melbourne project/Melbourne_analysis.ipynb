# Data Understanding

## Dataset: Melbourne House Prices Analysis


| Column | Description |
|---|---|
| `Suburb` | Name of the suburb where the property is located |
| `Address` | Street address of the property |
| `Rooms` | Number of rooms |
| `Type` | Property type: h (house, cottage, villa, terrace), u (unit, duplex), t (townhouse) |
| `Price` | Sale price in Australian dollars (AUD) |
| `Method` | Sale method: S (sold), SP (sold prior), PI (passed in), PN (sold prior, not disclosed), SN (sold, not disclosed), VB (vendor bid), W (withdrawn prior to auction), SA (sold after auction), SS (sold after auction, price not disclosed) |
| `SellerG` | Real estate agent or agency that sold the property |
| `Date` | Date the property was sold (text, format dd/mm/yyyy) |
| `Distance` | Distance from the Melbourne CBD (central business district) in km |
| `Postcode` | Postal code of the property |
| `Bedroom2` | Number of bedrooms, from a second source (it can differ from `Rooms`) |
| `Bathroom` | Number of bathrooms |
| `Car` | Number of car parking spots |
| `Landsize` | Land size in square meters |
| `BuildingArea` | Building size in square meters |
| `YearBuilt` | Year the building was built |
| `CouncilArea` | Governing council for the area |
| `Lattitude` | Latitude of the property (the column name is misspelled in the file) |
| `Longtitude` | Longitude of the property (also misspelled in the file) |
| `Regionname` | General region (West, North West, North, North East, and so on) |
| `Propertycount` | Number of properties in the suburb |

# Data Exploration and cleaning

```python
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv('MELBOURNE_HOUSE_PRICES_LESS.csv')
df
```

```python
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)
df['Postcode'] = df['Postcode'].astype(str)
```

### 7)  Handling Missing Values

```python
num_cols = df.select_dtypes(include= 'number').columns
num_cols
```

```python
df[num_cols]
```

```python
from sklearn.impute import KNNImputer

knn = KNNImputer(n_neighbors= 5)

df[num_cols] = knn.fit_transform(df[num_cols])
```

### 8)  Check outliers

#### 8.1 Detect Outliers with datasist

```python
from datasist.structdata import detect_outliers

outlier_indices = detect_outliers(data= df, n= 0, features= num_cols)
outlier_indices
```

```python
len(outlier_indices)
```

```python
len(outlier_indices)/len(df)*100
```

```python
for col in num_cols:
    outliers_count=len(detect_outliers(data= df, n= 0, features= [col]))
    percentage=outliers_count/len(df)*100
    print(col,":", outliers_count,"--->",round(percentage,2),"%")
```

# Data Analysis

###   What is the average property prices for every region ?

```python
successful_sales = cleaned_df[cleaned_df["Method"].isin(["S", "SP", "SA"])]
avg_by_region = (successful_sales.groupby('Regionname')['Price']
                   .mean()
                   .sort_values(ascending=False)
                   .reset_index())

avg_by_region.index = avg_by_region.index + 1
avg_by_region
```

```python
px.bar(data_frame=avg_by_region, x='Regionname', y='Price',
       color='Regionname', text_auto='.2s',
       title='Average Property Price by Region',labels={'Regionname': 'Region Name'})
```

###   What are the top 10 most expensive suburbs by average price?

```python
top10_expensive = (successful_sales.groupby('Suburb')['Price']
                   .mean()
                   .sort_values(ascending=False)
                   .head(10)
                   .reset_index())

top10_expensive .index = top10_expensive .index + 1
top10_expensive
```

```python
px.bar(data_frame=top10_expensive , x='Suburb', y='Price',
       color='Suburb', text_auto='.2s',
       title='Top 10 most expensive suburbs by average price')
```

###  What are the 10 cheapest suburbs by average price?

```python
top10_cheapest = (successful_sales.groupby('Suburb')['Price']
                    .mean()
                    .sort_values()
                    .head(10)
                    .reset_index())

top10_cheapest.index = top10_cheapest.index + 1
top10_cheapest
```

```python
px.bar(data_frame=top10_cheapest, x='Suburb', y='Price',
       color='Suburb', text_auto='.2s',
       title='Top 10 Cheapest Suburbs by Average Price')
```

### Which property type is sold the most?

```python
px.pie(data_frame=type_counts, names='Type', values='Count',title='Percentage of sold property types ')
```

```python
px.bar(data_frame=avg_price_by_type, x='Type', y='Average Price',
       color='Type', text_auto='.2s',
       title='Average Price by Property Type')
```

### Which sellers (agents) sold the most properties?

### Which year has the highest number of sales?

```python
sold_properties['Year'] = sold_properties['Date'].dt.year.astype(str)
sales_per_year = sold_properties['Year'].value_counts().reset_index()
sales_per_year.columns = ['Year', 'Sales Count']
sales_per_year.index=sales_per_year.index + 1
sales_per_year
```

```python
px.bar(sales_per_year,
    x='Year',
    y='Sales Count',
    text='Sales Count',  
    color='Year',        
    title='Total Number of Sales per Year',)
```

### What is the Annual revenue?

```python
successful_sales['Year'] = successful_sales['Date'].dt.year.astype(str)

Annual_revenue = (successful_sales.groupby('Year')['Price'].sum().reset_index())
Annual_revenue.columns = ['Year', 'Total Revenue']
Annual_revenue.index=Annual_revenue.index + 1 
Annual_revenue
```

```python
px.bar(Annual_revenue,
    x='Year',
    y='Total Revenue',
    text='Total Revenue',  
    color='Year', 
    text_auto=True ,
    title='Total Revenue per Year')
```

```python
px.line(data_frame=Annual_revenue, 
                    x='Year', 
                    y='Total Revenue', 
                    markers=True,
                    title='Annual Total Revenue')
```

### What is the monthly revenue?

```python
successful_sales['YearMonth'] = successful_sales['Date'].dt.strftime('%Y-%m')
monthly_revenue = successful_sales.groupby('YearMonth')['Price'].sum().reset_index()
monthly_revenue.index=monthly_revenue.index + 1
monthly_revenue
```

```python
px.line(data_frame=monthly_revenue, 
                    x='YearMonth', 
                    y='Price', 
                    markers=True,
                    labels={'YearMonth':'Month-Year' , 'Price':'Total Revenue'} ,
                    title='Monthly Total Revenue')
```

### What is the daily revenue?

```python
Daily_revenue = ( successful_sales.groupby('Date')['Price'] .sum().reset_index())
Daily_revenue.index=Daily_revenue.index + 1
Daily_revenue
```

```python
 px.line( data_frame=Daily_revenue,
    x='Date',
    y='Price',
    markers=True,
     labels={'Price':'Total Revenue'},
    title='Daily Total Revenue')
```

### Are properties closer to the city center more expensive?

```python
price_by_distance = successful_sales.groupby(['Distance','Type'])['Price'].median().reset_index()
price_by_distance.index=price_by_distance.index+1
price_by_distance.head(10)
```

```python
 px.scatter(data_frame=price_by_distance,
    x='Distance',  
    y='Price', 
    color='Type' ,    
    title='Property Price vs Distance from City Center')
```
